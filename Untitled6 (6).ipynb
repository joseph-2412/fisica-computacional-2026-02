## Parte B (y D): Módulo de Integración de Romberg (`modulo_romberg.f90`)


```python
%%writefile taller_oscilador_romberg/modulo_romberg.f90
module modulo_romberg
  implicit none
  ! Definición de precisión doble estándar
  integer, parameter :: dp = kind(1.0d0)

  ! Parámetros  del oscilador
  ! gamma = 0.5 s^-1 (coeficiente de amortiguamiento)
  ! omega = 2*pi rad/s (frecuencia angular amortiguada)
  real(dp), parameter :: gamma_param = 0.5_dp
  real(dp), parameter :: omega_param = 2.0_dp * 3.14159265358979323846_dp

contains

! Función: integrando_energia(t)
  real(dp) function integrando_energia(t)
    real(dp), intent(in) :: t
    real(dp) :: v
    v = -exp(-gamma_param * t) * (gamma_param * cos(omega_param * t) + omega_param * sin(omega_param * t))
    integrando_energia = 2.0_dp * gamma_param * (v**2)
  end function integrando_energia

!Función: integrando_pi(x)
  real(dp) function integrando_pi(x)
    real(dp), intent(in) :: x
    integrando_pi = 4.0_dp / (1.0_dp + x**2)
  end function integrando_pi

! Función: f_eval(opcion, x)
  real(dp) function f_eval(opcion, x)
    integer, intent(in) :: opcion
    real(dp), intent(in) :: x
    if (opcion == 1) then
       f_eval = integrando_energia(x)
    else
       f_eval = integrando_pi(x)
    end if
  end function f_eval

! Subrutina: romberg
! Calcula la integral definida en [a, b] construyendo la matriz triangular
! de Romberg S(k, j) hasta alcanzar |S(k,k) - S(k-1,k-1)| < tol o k = kmax.
  subroutine romberg(opcion, a, b, tol, kmax, S, k_final, n_eval)
    integer, intent(in) :: opcion, kmax
    real(dp), intent(in) :: a, b, tol
    real(dp), allocatable, intent(out) :: S(:,:)
    integer, intent(out) :: k_final, n_eval

    integer :: k, j, i, puntos_nuevos
    real(dp) :: h, suma, factor

    allocate(S(0:kmax, 0:kmax))
    S = 0.0_dp

    h = b - a
    S(0,0) = 0.5_dp * h * (f_eval(opcion, a) + f_eval(opcion, b))
    n_eval = 2
    k_final = 0

    do k = 1, kmax
       h = h / 2.0_dp
       puntos_nuevos = 2**(k - 1)
       suma = 0.0_dp
       do i = 1, puntos_nuevos
          suma = suma + f_eval(opcion, a + real(2*i - 1, dp) * h)
       end do
       n_eval = n_eval + puntos_nuevos
       S(k,0) = 0.5_dp * S(k-1,0) + h * suma

       do j = 1, k
          factor = 4.0_dp**j
          S(k,j) = S(k,j-1) + (S(k,j-1) - S(k-1,j-1)) / (factor - 1.0_dp)
       end do

       if (abs(S(k,k) - S(k-1,k-1)) < tol) then
          k_final = k
          exit
       end if
    end do

    if (k_final == 0) k_final = kmax
  end subroutine romberg

end module modulo_romberg
```

## Parte B: Cálculo de Energía Disipada (`programa_energia.f90`)


```python
%%writefile taller_oscilador_romberg/programa_energia.f90
! Programa:  programa_energia


program programa_energia
  use modulo_romberg, only: dp, romberg, gamma_param, omega_param
  implicit none
  real(dp) :: a, b, tol, e_dis, e_0, e_t, e_ref, dif, omega02, v0, vT, x0, xT
  integer :: kmax, k_final, n_eval, i, j
  real(dp), allocatable :: S(:,:)

! 1. Parámetros de integración en el intervalo [0, 3] s
  a = 0.0_dp
  b = 3.0_dp
  tol = 1.0e-6_dp
  kmax = 12

! 2. Llamar al integrador de Romberg (opcion = 1 -> potencia disipada)
  call romberg(1, a, b, tol, kmax, S, k_final, n_eval)
  e_dis = S(k_final, k_final)

! 3. Cálculo analítico de la energía mecánica total
  omega02 = omega_param**2 + gamma_param**2

  ! Energía mecánica inicial en t = 0
  x0 = exp(-gamma_param * a) * cos(omega_param * a)
  v0 = -exp(-gamma_param * a) * (gamma_param * cos(omega_param * a) + omega_param * sin(omega_param * a))
  e_0 = 0.5_dp * v0**2 + 0.5_dp * omega02 * x0**2

! Energía mecánica final en t = T = 3.0 s
  xT = exp(-gamma_param * b) * cos(omega_param * b)
  vT = -exp(-gamma_param * b) * (gamma_param * cos(omega_param * b) + omega_param * sin(omega_param * b))
  e_t = 0.5_dp * vT**2 + 0.5_dp * omega02 * xT**2

! 4. Diferencia exacta de energía y discrepancia numérica
  e_ref = e_0 - e_t
  dif = abs(e_dis - e_ref)

  ! Exportar tabla_energia.dat (filas k=0..k_final)
  open(unit=10, file='tabla_energia.dat', status='replace')
  do i = 0, k_final
     write(10, '(13(es24.16,1x))') (S(i, j), j=0, i)
  end do
  close(10)

  ! Exportar resultado_energia.dat
  open(unit=11, file='resultado_energia.dat', status='replace')
  write(11, '(i5,1x,i5,1x,es24.16,1x,es24.16,1x,es24.16)') &
        k_final, n_eval, e_dis, e_ref, dif
  close(11)

  deallocate(S)
end program programa_energia
```

## Parte C: Derivación Numérica del Oscilador (`derivadas_oscilador.py`)


```python
%%writefile taller_oscilador_romberg/derivadas_oscilador.py
import numpy as np
import matplotlib.pyplot as plt

# Constantes del oscilador
G, W = 0.5, 2 * np.pi

def x_ex(t): return np.exp(-G * t) * np.cos(W * t)
def v_ex(t): return -np.exp(-G * t) * (G * np.cos(W * t) + W * np.sin(W * t))
def a_ex(t): return np.exp(-G * t) * ((G**2 - W**2) * np.cos(W * t) + 2 * G * W * np.sin(W * t))

def calcular_derivadas(dt):
    n_puntos = int(round(3.0 / dt)) + 1
    t = np.linspace(0.0, 3.0, n_puntos)
    x = x_ex(t)

    # 1. Primera derivada (velocidad) por rebanadas
    v_adel = (x[1:] - x[:-1]) / dt
    v_atra = (x[1:] - x[:-1]) / dt
    v_cent = (x[2:] - x[:-2]) / (2 * dt)
    v_cinco = (-x[4:] + 8*x[3:-1] - 8*x[1:-3] + x[:-4]) / (12 * dt)

    # 2. Segunda derivada (aceleración) central
    a_cent = (x[2:] - 2*x[1:-1] + x[:-2]) / (dt**2)

    # Errores máximos frente a valores exactos
    err_v_adel = np.max(np.abs(v_adel - v_ex(t[:-1])))
    err_v_atra = np.max(np.abs(v_atra - v_ex(t[1:])))
    err_v_cent = np.max(np.abs(v_cent - v_ex(t[1:-1])))
    err_v_cinco = np.max(np.abs(v_cinco - v_ex(t[2:-2])))
    err_a_cent = np.max(np.abs(a_cent - a_ex(t[1:-1])))

    return t, x, v_cent, a_cent, [err_v_adel, err_v_atra, err_v_cent, err_v_cinco, err_a_cent]

# Ejecución para dt = 0.01 y dt = 0.005
t_01, x_01, v_c01, a_c01, errs_01 = calcular_derivadas(0.01)
_, _, _, _, errs_005 = calcular_derivadas(0.005)

# Guardar derivadas_oscilador.dat con doble precisión completa (16 decimales)
datos_interiores = np.column_stack((t_01[1:-1], x_01[1:-1], v_c01, a_c01))
np.savetxt("derivadas_oscilador.dat", datos_interiores, fmt="%.16e")

# Guardar errores_derivadas.dat
tabla_errores = np.array([
    [0.01, *errs_01],
    [0.005, *errs_005]
])
np.savetxt("errores_derivadas.dat", tabla_errores, fmt="%.16e")

# Generar gráfica derivadas_oscilador.png
t_int = t_01[1:-1]
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)

ax1.plot(t_int, v_c01, 'b-', label='v num (central)', alpha=0.8)
ax1.plot(t_int, v_ex(t_int), 'r--', label='v exacta', alpha=0.7)
ax1.set_ylabel('Velocidad [m/s]')
ax1.legend(loc='upper right')
ax1.grid(True, alpha=0.3)

ax2.plot(t_int, a_c01, 'g-', label='a num (central)', alpha=0.8)
ax2.plot(t_int, a_ex(t_int), 'm--', label='a exacta', alpha=0.7)
ax2.set_xlabel('Tiempo [s]')
ax2.set_ylabel('Aceleración [m/s²]')
ax2.legend(loc='upper right')
ax2.grid(True, alpha=0.3)

plt.suptitle(r'Oscilador amortiguado: Velocidad y Aceleración ($\Delta t = 0.01$ s)')
plt.tight_layout()
plt.savefig('derivadas_oscilador.png', dpi=150)
plt.close()
```

## Parte D: Validación del Módulo con el Cálculo de $\pi$ (`programa_pi.f90`)


```python
%%writefile taller_oscilador_romberg/programa_pi.f90

! Programa:  programa_pi
! Propósito: Calcular pi = int_0^1 4/(1+x^2) dx usando modulo_romberg

program programa_pi
  use modulo_romberg, only: dp, romberg
  implicit none
  real(dp), parameter :: pi_exacto = 3.14159265358979323846_dp
  real(dp) :: a, b, tol, res_pi, error_abs
  integer :: kmax, k_final, n_eval
  real(dp), allocatable :: S(:,:)

! 1. Límites de integración [0, 1] y tolerancia requerida (10^-5)
  a = 0.0_dp
  b = 1.0_dp
  tol = 1.0e-5_dp
  kmax = 10

! 2. Invocar Romberg seleccionando opcion = 2 (integrando de pi)
  call romberg(2, a, b, tol, kmax, S, k_final, n_eval)
  res_pi = S(k_final, k_final)
  error_abs = abs(res_pi - pi_exacto)

! 3. Guardar nivel alcanzado, evaluaciones totales, valor de pi y error absoluto
  open(unit=12, file='resultado_pi.dat', status='replace')
  write(12, '(i5,1x,i5,1x,es24.16,1x,es24.16)') &
        k_final, n_eval, res_pi, error_abs
  close(12)

  deallocate(S)
end program programa_pi
```

```python
# ===================== VERIFICACIÓN DEL TALLER 7 (no modificar) =====================
import os
import numpy as np

def actualizado(dat, fuentes):
    """El archivo existe y es posterior a todos los códigos fuente."""
    if not os.path.exists(dat):
        return False, f"no existe {dat}"
    for f in fuentes:
        if not os.path.exists(f):
            return False, f"no existe {f}"
        if os.path.getmtime(dat) < os.path.getmtime(f):
            return False, f"{dat} es anterior a {f}: vuelva a ejecutar el programa"
    return True, ""

G, W = 0.5, 2*np.pi
def x_ex(t): return np.exp(-G*t)*np.cos(W*t)
def v_ex(t): return -np.exp(-G*t)*(G*np.cos(W*t) + W*np.sin(W*t))
def a_ex(t): return np.exp(-G*t)*((G*G - W*W)*np.cos(W*t) + 2*G*W*np.sin(W*t))
def energia(t): return 0.5*v_ex(t)**2 + 0.5*(W*W + G*G)*x_ex(t)**2

resultados = []
def caso(nombre, funcion):
    try:
        ok, detalle = funcion()
    except Exception as e:
        ok, detalle = False, f"error al leer: {e}"
    resultados.append(ok)
    print(f"[{'PASS' if ok else 'FAIL'}] {nombre}" + (f"  -> {detalle}" if detalle else ""))

def c1():
    ok, d = actualizado("resultado_energia.dat", ["modulo_romberg.f90", "programa_energia.f90"])
    if not ok: return ok, d
    k, n, e_dis, e_ref_est, dif = np.loadtxt("resultado_energia.dat")
    e_ref = energia(0.0) - energia(3.0)
    if abs(e_ref_est - e_ref) > 1e-10: return False, f"E(0) - E(T) = {e_ref_est:.10f}, se espera {e_ref:.10f}"
    if abs(e_dis - e_ref) >= 1e-6: return False, f"|E_dis - (E(0) - E(T))| = {abs(e_dis - e_ref):.2e}"
    if int(k) != 8 or int(n) != 257: return False, f"k_final = {int(k)}, evaluaciones = {int(n)} (se esperan 8 y 257)"
    return True, f"E_dis = {e_dis:.10f} J/kg, k = 8, 257 evaluaciones"

def c2():
    ok, d = actualizado("derivadas_oscilador.dat", ["derivadas_oscilador.py"])
    if not ok: return ok, d
    t, x, v, a = np.loadtxt("derivadas_oscilador.dat", unpack=True)
    if t.size != 299 or abs(t[1] - t[0] - 0.01) > 1e-12: return False, "se esperan 299 puntos interiores con dt = 0.01"
    if np.max(np.abs(x - x_ex(t))) > 1e-12: return False, "las posiciones no corresponden al oscilador"
    ev, ea = np.abs(v - v_ex(t)).max(), np.abs(a - a_ex(t)).max()
    if not (0.9*3.765e-3 <= ev <= 1.1*3.765e-3): return False, f"error de v = {ev:.3e}"
    if not (0.9*1.215e-2 <= ea <= 1.1*1.215e-2): return False, f"error de a = {ea:.3e}"
    return True, f"errores máximos v {ev:.3e}, a {ea:.3e}"

def c3():
    ok, d = actualizado("errores_derivadas.dat", ["derivadas_oscilador.py"])
    if not ok: return ok, d
    e = np.loadtxt("errores_derivadas.dat")
    if e.shape != (2, 6): return False, f"forma {e.shape}, se espera (2, 6)"
    if abs(e[0, 4] - 3.0275e-6)/3.0275e-6 > 0.1: return False, f"error de cinco puntos con dt = 0.01: {e[0,4]:.3e}"
    r = e[0, 1:]/e[1, 1:]
    rangos = [(1.9, 2.1), (1.9, 2.1), (3.8, 4.2), (15.0, 17.0), (3.7, 4.2)]
    if not all(lo <= ri <= hi for ri, (lo, hi) in zip(r, rangos)):
        return False, "razones " + ", ".join(f"{ri:.2f}" for ri in r)
    return True, "razones " + ", ".join(f"{ri:.2f}" for ri in r)

def c4():
    ok, d = actualizado("resultado_pi.dat", ["modulo_romberg.f90", "programa_pi.f90"])
    if not ok: return ok, d
    k, n, r, e = np.loadtxt("resultado_pi.dat")
    if abs(r - np.pi) >= 1e-5: return False, f"|R - pi| = {abs(r - np.pi):.2e}"
    if int(k) != 4 or int(n) != 17: return False, f"k_final = {int(k)}, evaluaciones = {int(n)} (se esperan 4 y 17)"
    return True, f"pi = {r:.10f}, k = 4, 17 evaluaciones"

caso("1. Energía disipada: Romberg con tol 1e-6 coincide con E(0) - E(T)", c1)
caso("2. Velocidad y aceleración centrales con dt = 0.01", c2)
caso("3. Órdenes de las cinco fórmulas de derivación", c3)
caso("4. pi con el mismo módulo: tol 1e-5, 17 evaluaciones", c4)
print(f"\nResultado: {sum(resultados)} de {len(resultados)} PASS")
```
